Compare PTHA curves with the given data

```python
def exceedance_curve(thresholds, max_stage_point, rates): 
   # Annual Mean Rate of threshold exceedance 
   lambda_exc = np.zeros((len(thresholds), 1000)) 
   for threshold in range(len(thresholds)): 
       ix = np.array(np.where(max_stage_point > thresholds[threshold])).squeeze() 
       for j in range(1000): 
           #for index in ix: 
           lambda_exc[threshold, j] = lambda_exc[threshold, j] +  rates[ix,j].sum(axis=0) 
   #------------------------------------------------ 
   return lambda_exc 

def single_exceedance_curve(thresholds, max_stage_point, rates): 
   # Annual Mean Rate of threshold exceedance 
   lambda_exc = np.zeros((len(thresholds), )) 
   for threshold in range(len(thresholds)): 
       ix = np.array(np.where(max_stage_point > thresholds[threshold])).squeeze() 
       #for index in ix: 
       lambda_exc[threshold] = lambda_exc[threshold] +  rates[ix].sum(axis=0) 
   #------------------------------------------------ 
   return lambda_exc
```

```python
eve_perf = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_compile_combined.csv')

task = 'loss_func_MSE'
eve_dep_mse = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

task = 'mse_asym2'
eve_dep_asym2 = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

task = 'mse_asym5'
eve_dep_asym5 = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

task = 'mse_asym10'
eve_dep_asym10 = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

task = 'mse_asym15'
eve_dep_asym15 = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

task = 'mse_asym20'
eve_dep_asym20 = pd.read_csv(f'{MLDir}/model/{reg}/{task}/out/model_direct_off[64, 128, 256]_on[16, 128, 128]_{train_size}_true_pred_er_combined.csv')

#iterate over points
for select_pt in range(1, 12):
    select_pt = str(select_pt)
    true = eve_dep_mse['T'+select_pt].values
    pred = eve_dep_mse['P'+select_pt].values
    pred_asym2 = eve_dep_asym2['P'+select_pt].values
    pred_asym5 = eve_dep_asym5['P'+select_pt].values
    pred_asym10 = eve_dep_asym10['P'+select_pt].values
    pred_asym15 = eve_dep_asym15['P'+select_pt].values
    pred_asym20 = eve_dep_asym20['P'+select_pt].values
    rate = eve_perf['mean_prob'].values

    #compute depth thresholds 
    thresholds_depth = np.linspace(0, 20, 100)

    #compute exceedance curve for true
    lambda_exc_true = single_exceedance_curve(thresholds_depth, true, rate)
    lambda_exc_pred = single_exceedance_curve(thresholds_depth, pred, rate)
    lambda_exc_pred_asym2 = single_exceedance_curve(thresholds_depth, pred_asym2, rate)
    lambda_exc_pred_asym5 = single_exceedance_curve(thresholds_depth, pred_asym5,rate)
    lambda_exc_pred_asym10 = single_exceedance_curve(thresholds_depth, pred_asym10, rate)
    lambda_exc_pred_asym15 = single_exceedance_curve(thresholds_depth, pred_asym15, rate)
    lambda_exc_pred_asym20 = single_exceedance_curve(thresholds_depth, pred_asym20, rate)

    #plot exceedance curve
    plt.figure()
    plt.plot(thresholds_depth,lambda_exc_true, label='True', color='black', linestyle='dashed')
    plt.plot(thresholds_depth,lambda_exc_pred, label='Pred_MSE', color='blue')
    # plt.plot(thresholds_depth,lambda_exc_pred_asym2, label='Pred_Asym2', color='pink')
    # plt.plot(thresholds_depth,lambda_exc_pred_asym5, label='Pred_Asym5', color='red')
    # plt.plot(thresholds_depth,lambda_exc_pred_asym10, label='Pred_Asym10', color='orange')
    plt.plot(thresholds_depth,lambda_exc_pred_asym15, label='Pred_Asym15', color='purple')
    plt.plot(thresholds_depth,lambda_exc_pred_asym20, label='Pred_Asym20', color='green')

    plt.xlabel('Depth')
    plt.ylabel('Exceedance rate')
    plt.yscale('log')
    # plt.xscale('log')
    #fix y axis limits
    # plt.ylim(1e-6, 1)
    plt.xlim(0.1, 12)
    plt.legend()
    plt.title(f'Exceedance curve for gauge {select_pt}')
    plt.show()
```
